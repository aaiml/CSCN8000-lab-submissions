```python
#create two matrices
e1 = np.array([[5,7,2],[4,5,6],[7,4,2]])
e2 = np.array([[4,2],[6,2],[4,2]])

#it is possible when the 3x3 is multiplied onto the 2x3.
e3 = np.matmul(e1,e2)
print(e3)

#it is not possible when the 2x3 is multiplied onto the 3x3. matrix multiplication is not commutative.
try:
    e4 = np.matmul(e2,e1)
except ValueError as msg:
    print(msg)
```

```python
# Broadcasting applies a 1D array to each row of a compatible 3x3 matrix.
g1 = np.array([[10,20,30],
                [40,50,60],
                [70,80,90]])
g2 = np.array([1,2,3])

# The vector is broadcast across all three rows.
g3 = g1 - g2
g4 = g1 * g2

print("Subtraction:")
print(g3)
print("Element-wise multiplication:")
print(g4)
```

```python
#solve the following two system of linear equations using matrices (find the values of x1 and x2) and determine the number of solutions using the functions.
# 2x1 +3x2 - 4x3 = 6, x1-4x2 = 8
#3y1-4y2+5y3=10, -y1+2y2-4y3=8
h1 = np.array([[2,3,-4],[1,-4,0]])
h2 = np.array([6,8])

h3 = np.linalg.lstsq(h1,h2)[0]

h4 = np.array([[3,-4,5],[-1,2,-4]])
h5 = np.array([10,8])

h6 = np.linalg.lstsq(h4,h5)[0]

print(h3)
print(h6)
```
