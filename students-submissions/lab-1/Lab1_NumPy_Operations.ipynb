 The `np.random.rand()` function is used to generate random numbers between 0 and 1. In this example, an array containing three random values is created. The `shape` attribute is used to display the shape of the array.

 A4. 2-D Array and Slicing

A 2-D NumPy array is created using the given values. Array slicing is then used to select specific rows. The first two rows are selected using `[:1]`, while the last two rows are selected using `[-1:]`.

```python
array_2d = np.array([
    [10, 20, 45],
    [30, 12, 16],
    [42, 17, 56]
])

print("2-D Array:")
print(array_2d)

print("\nFirst one rows:")
print(array_2d[:1])

print("\nLast one rows:")
print(array_2d[-1:])
```

A5. Stacking and Splitting

Two 2 × 2 arrays are created and combined using NumPy stacking operations. `np.vstack()` combines the arrays vertically, while `np.hstack()` combines them horizontally. The `np.vsplit()` function is used to divide the combined array into smaller arrays.

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("A:")
print(A)

print("\nB:")
print(B)

print("\nVertical Stack:")
print(np.vstack((A, B)))

print("\nHorizontal Stack:")
print(np.hstack((A, B)))

print("\nSplit:")
for part in np.vsplit(np.vstack((A, B)), 2):
    print(part)
```

A6. Matrix Multiplication


Matrix multiplication is possible when the number of columns in the first matrix equals the number of rows in the second matrix. X has shape (3,3) and Y has shape (3,2), so the multiplication is possible.

```python
X = np.array([
    [5, 7, 2],
    [4, 5, 6],
    [7, 4, 1]
])

Y = np.array([
    [4, 2],
    [6, 2],
    [4, 3]
])

print("Shape of X:", X.shape)
print("Shape of Y:", Y.shape)

print("\nX × Y:")
print(np.matmul(X, Y))
```

A7. Vector Shape and Dimensions

The vector x is created and its shape and number of dimensions are determined using the `shape` and `ndim` attributes. The vector is then reshaped into a matrix with shape (3, 1). The same reshaping operation is performed on y, changing each vector from one dimension to a two-dimensional matrix.

```python
x = np.array([2, -1, -8])
y = np.array([3, 1, -2])

print("x:", x)
print("Shape of x:", x.shape)
print("Dimensions of x:", x.ndim)

mx = x.reshape(3, 1)

print("\nReshaped x:")
print(mx)
print("Shape:", mx.shape)
print("Dimensions:", mx.ndim)

my = y.reshape(3, 1)

print("\nReshaped y:")
print(my)
print("Shape:", my.shape)
print("Dimensions:", my.ndim)
```

 A8. Broadcasting

 Broadcasting allows NumPy to perform arithmetic operations between arrays with compatible shapes without manually repeating the smaller array. In this example, a 1-D vector is combined with a 3 × 3 matrix. NumPy automatically applies the vector to each row of the matrix for both subtraction and multiplication.

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

vector = np.array([1, 2, 3])

print("Matrix:")
print(matrix)

print("\nVector:")
print(vector)

print("\nBroadcasting subtraction:")
print(matrix - vector)

print("\nBroadcasting multiplication:")
print(matrix * vector)
```

Part B

In this part, two systems of linear equations are represented using matrices. The coefficient matrix and the augmented matrix are used to determine the number of solutions. The matrix rank is compared with the number of variables to determine whether the system has a unique solution, no solution, or infinitely many solutions.

B1. System of Linear Equations

The first system contains two equations and three unknown variables. The coefficient matrix and augmented matrix are created using NumPy. Their ranks are calculated to determine the number of solutions. Since the ranks are equal and less than the number of variables, the system is consistent and has infinitely many solutions. The `linsolve()` function is used to display the solution in parametric form.

```python
import sympy as sp
```

```python
A1 = np.array([
    [2, 3, -4],
    [1, -4, 0]
], dtype=float)

b1 = np.array([6, 8], dtype=float)

augmented1 = np.column_stack((A1, b1))

print("Coefficient Matrix A:")
print(A1)

print("\nConstants b:")
print(b1)

print("\nRank of A:", np.linalg.matrix_rank(A1))
print("Rank of augmented matrix:",
      np.linalg.matrix_rank(augmented1))
```

```python
x1, x2, x3 = sp.symbols('x1 x2 x3')

equations1 = [
    sp.Eq(2*x1 + 3*x2 - 4*x3, 6),
    sp.Eq(x1 - 4*x2, 8)
]

solution1 = sp.linsolve(equations1, (x1, x2, x3))

print("Solution:")
print(solution1)
```

B2. Number of Solutions

The second system also contains two equations and three unknown variables. The ranks of the coefficient matrix and augmented matrix are calculated and compared. Since the ranks are equal and less than the number of variables, the system has infinitely many solutions. The `linsolve()` function is used to find and display the parametric solution.



```python
A2 = np.array([
    [3, -4, 5],
    [-1, 2, -4]
], dtype=float)

b2 = np.array([10, 8], dtype=float)

augmented2 = np.column_stack((A2, b2))

print("Coefficient Matrix A:")
print(A2)

print("\nConstants b:")
print(b2)

print("\nRank of A:", np.linalg.matrix_rank(A2))
print("Rank of augmented matrix:",
      np.linalg.matrix_rank(augmented2))
```

```python
y1, y2, y3 = sp.symbols('y1 y2 y3')

equations2 = [
    sp.Eq(3*y1 - 4*y2 + 5*y3, 10),
    sp.Eq(-y1 + 2*y2 - 4*y3, 8)
]

solution2 = sp.linsolve(equations2, (y1, y2, y3))

print("Solution:")
print(solution2)
```
