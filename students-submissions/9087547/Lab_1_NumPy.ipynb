### Broadcasting Explanation

Broadcasting allows NumPy to perform operations between arrays with compatible shapes.

In this example, the vector `[1, 2, 3]` is automatically applied to each row of the 3 × 3 matrix.

For subtraction, NumPy subtracts 1, 2, and 3 from each row.

For multiplication, NumPy multiplies each column by 1, 2, and 3 respectively.

```python
A1 = np.array([
    [2, 3, -4],
    [1, -4, 0]
])

b1 = np.array([6, 8])

print("Coefficient matrix A1:")
print(A1)

print("\nConstant vector b1:")
print(b1)

print("\nShape of A1:", A1.shape)
```

```python
rank_A1 = np.linalg.matrix_rank(A1)

augmented_A1 = np.column_stack((A1, b1))
rank_augmented_A1 = np.linalg.matrix_rank(augmented_A1)

print("Rank of A1:", rank_A1)
print("Rank of augmented matrix:", rank_augmented_A1)
```

```python
if rank_A1 == rank_augmented_A1:
    if rank_A1 == A1.shape[1]:
        print("The system has one unique solution.")
    else:
        print("The system has infinitely many solutions.")
else:
    print("The system has no solution.")
```

```python
solution1, residuals1, rank1, singular_values1 = np.linalg.lstsq(A1, b1, rcond=None)

print("One least-squares solution:")
print(solution1)

print("Rank:", rank1)
print("Residuals:", residuals1)
```

```python
A2 = np.array([
    [3, -4, 5],
    [-1, 2, -4]
])

b2 = np.array([10, 8])

print("Coefficient matrix A2:")
print(A2)

print("\nConstant vector b2:")
print(b2)
```
