### 4. Create a 2 D array [[10,20,45], [30,12,16], [42,17,56]] and perform the following operations: 
     
       Slice the 2D array to get the first two rows, slice the 2D array to get the last two rows.

```python
# Create the 2D array (3 rows, 3 columns)
matrix_2d = np.array([[10, 20, 45], 
                      [30, 12, 16], 
                      [42, 17, 56]])

print("Original 2D Array:\n", matrix_2d)

# Slice to get the first two rows (rows 0 and 1)
first_two_rows = matrix_2d[:2, :]
print("\nFirst two rows:\n", first_two_rows)

# Slice to get the last two rows (rows 1 and 2)
last_two_rows = matrix_2d[-2:, :]
print("\nLast two rows:\n", last_two_rows)
```

### 5. Create two 2x2 arrays and demonstrate how you can stack the elements vertically, horizontally, and split the arrays into smaller arrays.

```python
# Define two 2x2 arrays
A = np.array([[1, 2], 
              [3, 4]])

B = np.array([[5, 6], 
              [7, 8]])

# 1. Vertical Stacking (stacks rows on top of each other -> 4x2 result)
v_stacked = np.vstack((A, B))
print("Vertically Stacked (vstack):\n", v_stacked)

# 2. Horizontal Stacking (stacks columns side by side -> 2x4 result)
h_stacked = np.hstack((A, B))
print("\nHorizontally Stacked (hstack):\n", h_stacked)

# 3. Splitting arrays back into smaller arrays
# Split the 4x2 vertical stack into two 2x2 arrays along rows (axis=0)
split_parts = np.vsplit(v_stacked, 2)
print("\nFirst split half:\n", split_parts[0])
print("\nSecond split half:\n", split_parts[1])
```

### 7. Create two arrays, x = ([2, -1, -8]) y = ([3, 1, -2]), Find the Shape, Number of dimensions of vector x.
     
      Reshape the vector x to a matrix of size (3,1) and determine the number of dimensions after reshaping y to a matrix of (3,1).

```python
# Define vectors x and y
x = np.array([2, -1, -8])
y = np.array([3, 1, -2])

# Find shape and dimensions of vector x
print("Vector x:", x)
print("Shape of x:", x.shape)
print("Number of dimensions of vector x:", x.ndim)  # Output: 1

# Reshape vector x to size (3, 1)
x_reshaped = x.reshape(3, 1)
print("\nReshaped x (3x1 matrix):\n", x_reshaped)
print("New shape of x:", x_reshaped.shape)
print("Number of dimensions of x after reshape:", x_reshaped.ndim)  # Output: 2

# Reshape y to (3, 1) and determine its dimensions
y_reshaped = y.reshape(3, 1)
print("\nReshaped y (3x1 matrix):\n", y_reshaped)
print("Number of dimensions of y after reshape:", y_reshaped.ndim)  # Output: 2
```

### 8. How does broadcasting work? Demonstrate the subtraction, multiplication by considering a 3 x 3 matrix.

What is Broadcasting? NumPy allows arithmetic operations between arrays of different shapes by automatically "stretching" or duplicating the smaller array across the larger array so their shapes match, without making extra copies in memory.

```python
# Consider a 3x3 matrix
M = np.array([[10, 20, 30], 
              [40, 50, 60], 
              [70, 80, 90]])

# 1. Broadcasting with a Scalar (Single number)
# The scalar 5 is stretched across all 9 elements
print("Original 3x3 Matrix M:\n", M)
print("\nSubtraction (M - 5):\n", M - 5)
print("\nMultiplication (M * 2):\n", M * 2)

# 2. Broadcasting with a 1D Row Vector (Shape: 1x3)
row_vector = np.array([1, 2, 3])
# row_vector is broadcast across each of the 3 rows
print("\nBroadcasting Subtraction (M - [1, 2, 3]):\n", M - row_vector)
print("\nBroadcasting Multiplication (M * [1, 2, 3]):\n", M * row_vector)
```

### 1.Solve the following two system of linear equations using matrices (find the values of x1 and x2) and 
      determine the number of solutions using the functions. 
         1.2x1+3x2-4x3=6,x1-4x2=8
	
         2. 3y1-4y2+5y3=10,-y1+2y2-4y3=8

```python
# System 1:
# Equation 1: 2*x1 + 3*x2 - 4*x3 = 6
# Equation 2: 1*x1 - 4*x2 + 0*x3 = 8

# Matrix of coefficients (2 equations, 3 unknowns: x1, x2, x3)
A1 = np.array([[2, 3, -4], 
               [1, -4, 0]])

# Right-hand side constants
b1 = np.array([6, 8])

# Determine ranks to find the number of solutions
rank_A1 = np.linalg.matrix_rank(A1)
augmented_1 = np.column_stack((A1, b1))
rank_Aug1 = np.linalg.matrix_rank(augmented_1)
num_variables_1 = A1.shape[1]

print("=== SYSTEM 1 ===")
print("Rank of Coefficient Matrix A1:", rank_A1)
print("Rank of Augmented Matrix:", rank_Aug1)
print("Number of Variables:", num_variables_1)

# Check number of solutions using the rank test
if rank_A1 != rank_Aug1:
    print("Conclusion: No solution (Inconsistent system).")
elif rank_A1 == num_variables_1:
    print("Conclusion: Exactly 1 unique solution.")
else:
    print("Conclusion: INFINITELY MANY SOLUTIONS (Underdetermined: 2 equations, 3 unknowns).")

# Solve for the minimum-norm particular solution using least squares
sol1, residuals, rank, s = np.linalg.lstsq(A1, b1, rcond=None)
print("\nOne valid particular solution [x1, x2, x3]:")
print(f"x1 = {sol1[0]:.4f}")
print(f"x2 = {sol1[1]:.4f}")
print(f"x3 = {sol1[2]:.4f}")
```

```python
# System 2:
# Equation 1:  3*y1 - 4*y2 + 5*y3 = 10
# Equation 2: -1*y1 + 2*y2 - 4*y3 = 8

# Matrix of coefficients (2 equations, 3 unknowns: y1, y2, y3)
A2 = np.array([[3, -4, 5], 
               [-1, 2, -4]])

# Right-hand side constants
b2 = np.array([10, 8])

# Determine ranks to find number of solutions
rank_A2 = np.linalg.matrix_rank(A2)
augmented_2 = np.column_stack((A2, b2))
rank_Aug2 = np.linalg.matrix_rank(augmented_2)
num_variables_2 = A2.shape[1]

print("=== SYSTEM 2 ===")
print("Rank of Coefficient Matrix A2:", rank_A2)
print("Rank of Augmented Matrix:", rank_Aug2)
print("Number of Variables:", num_variables_2)

# Check number of solutions using the rank test
if rank_A2 != rank_Aug2:
    print("Conclusion: No solution (Inconsistent system).")
elif rank_A2 == num_variables_2:
    print("Conclusion: Exactly 1 unique solution.")
else:
    print("Conclusion: INFINITELY MANY SOLUTIONS (Underdetermined: 2 equations, 3 unknowns).")

# Solve for the minimum-norm particular solution using least squares
sol2, residuals, rank, s = np.linalg.lstsq(A2, b2, rcond=None)
print("\nOne valid particular solution [y1, y2, y3]:")
print(f"y1 = {sol2[0]:.4f}")
print(f"y2 = {sol2[1]:.4f}")
print(f"y3 = {sol2[2]:.4f}")
```
