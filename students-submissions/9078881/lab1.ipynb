# NumPy Operations

This notebook demonstrates NumPy array operations, indexing and slicing, reshaping, stacking, broadcasting, matrix multiplication, and solving systems of linear equations.


## Part A: NumPy Operations

### 1. An array from 1 to 20, incremented by 3

```python
import numpy as np

array1 = np.arange(1, 21, 3)
print(array1)
```

### 2. An array of 3 random numbers between 0 and 1

```python
random_array = np.random.rand(3)
print(random_array)
```

### 3. A 2-D array

```python
array2d = np.array([
    [10, 20, 45],
    [30, 12, 16],
    [42, 17, 56]
])

print("Original array:")
print(array2d)

print("\nFirst two rows:")
print(array2d[:2])

print("\nLast two rows:")
print(array2d[-2:])
```

### 4. A two 2×2 arrays

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("A:")
print(A)

print("\nB:")
print(B)

print("\nVertical stack:")
print(np.vstack((A, B)))

print("\nHorizontal stack:")
print(np.hstack((A, B)))

print("\nSplit the vertically stacked array:")
print(np.vsplit(np.vstack((A, B)), 2))
```

### 5. Matrix multiplication

```python
X = np.array([
    [5, 7, 2],
    [4, 5, 6],
    [7, 4, 2]
])

Y = np.array([
    [4, 2],
    [6, 2],
    [4, 2]
])

print("X shape:", X.shape)
print("Y shape:", Y.shape)

# X is 3x3 and Y is 3x2.
# The inner dimensions are both 3, so X @ Y is possible.

XY = X @ Y
print("\nX @ Y:")
print(XY)

# Y @ X is not possible because the inner dimensions are 2 and 3.
print("\nY @ X is not possible because the dimensions are (3,2) and (3,3).")
```

### 6. Shape and dimensions of vectors

```python
x = np.array([2, -1, -8])
y = np.array([3, 1, -2])

print("Shape of x:", x.shape)
print("Number of dimensions of x:", x.ndim)

x_matrix = x.reshape(3, 1)
y_matrix = y.reshape(3, 1)

print("\nx reshaped to (3, 1):")
print(x_matrix)

print("\ny reshaped to (3, 1):")
print(y_matrix)

print("\nNumber of dimensions of y after reshaping:", y_matrix.ndim)
```

### 7. Broadcasting Demonstration


```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

values = np.array([1, 2, 3])

print("Matrix:")
print(matrix)

print("\nValues:")
print(values)

print("\nSubtraction:")
print(matrix - values)

print("\nMultiplication:")
print(matrix * values)
```

## Part B: Systems of Linear Equations

### 1. System1

```python
A1 = np.array([[2, 3, -4], [1, -4, 0]], dtype=float)
b1 = np.array([6, 8], dtype=float)

print("Coefficient matrix A:")
print(A1)
print("\nConstants vector b:")
print(b1)
```

```python
# Let x3 be the free variable.
x3 = 0

x2 = (4 * x3 - 10) / 11
x1 = (48 + 16 * x3) / 11

print("For x3 =", x3)
print("x1 =", x1)
print("x2 =", x2)
print("x3 =", x3)
```

#### The number of solutions for system1 using NumPy functions

```python
rank_A1 = np.linalg.matrix_rank(A1)
rank_aug1 = np.linalg.matrix_rank(np.column_stack((A1, b1)))

print("Rank of coefficient matrix:", rank_A1)
print("Rank of augmented matrix:", rank_aug1)

if rank_A1 != rank_aug1:
    print("The system has no solution.")
elif rank_A1 == A1.shape[1]:
    print("The system has one unique solution.")
else:
    print("The system has infinitely many solutions.")
```
